```python
#bai3
a,b,c=map(int,input().split())
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print('tam giac deu')
    elif a==b or a==c or b==c:
        print('tam giac can')
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or c**2+b**2==a**2:
        print('tam giac vuong')
    else:
        print('tam giac thuong')
else: 
    print('khong phai la mot tam giac')
```
